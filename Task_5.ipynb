```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import seaborn as sns
```

```python
data = pd.read_csv('/content/Titanic_Cleaned_Dataset.csv')
df=data
df.head()
```

```python
df.isnull().sum()
df.fillna(0,inplace=True)
df
```

```python
df.isna().sum()
df
```

```python
#df['Sex'] = df['Sex'].map({'male':0, 'female':1})
x = df[['Sex', 'Age', 'Parch', 'Fare', 'SibSp', 'Pclass']]
y = df[['Survived']]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=40)
```

```python
df.head()
```

```python
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
principal=PCA(n_components=2)
z=principal.fit_transform(x_train)
xy=principal.transform(x_test)
```

```python
gh=[]
for i in range(1,11):
  model=KMeans(n_clusters=i,random_state=40)
  model.fit(z)
  gh.append(model.inertia_)
plt.plot(range(1,11),gh)
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.show()
```

```python
model_kmeans = KMeans(n_clusters=2, random_state=40, n_init='auto')
y_kmeans = model_kmeans.fit_predict(z)
```

```python
plt.scatter(z[y_kmeans==0,0], z[y_kmeans==0,1], s=100, c='magenta', label='Cluster 1')
plt.scatter(z[y_kmeans==1,0], z[y_kmeans==1,1], s=100, c='blue', label='Cluster 2')
#plt.scatter(z[y_kmeans==2,0], z[y_kmeans==2,1], s=100, c='orange', label='Cluster 3')
plt.scatter(model_kmeans.cluster_centers_[:,0], model_kmeans.cluster_centers_[:,1], s=200, c='black', label='Centroids')
plt.title('Clusters of Passengers (PCA Components)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()

```

```python
plt.title('PCA')
plt.xlabel('Pclass')
plt.ylabel('Age')
plt.scatter(z[:,0],z[:,1])
plt.show()
```

```python
print('Silhouette Score', silhouette_score(z,y_kmeans))
```

```python
from sklearn.metrics import accuracy_score
clusters = y_kmeans
acc1 = accuracy_score(y_train['Survived'], clusters)
acc2 = accuracy_score(y_train['Survived'], 1 - clusters)
accuracy = max(acc1, acc2)
print("Accuracy:", accuracy * 100)
```

```python
y_kmeans_test = model_kmeans.predict(xy)
predicted_clusters_df = pd.DataFrame(xy, columns=['Principal Component 1', 'Principal Component 2'])
predicted_clusters_df['Predicted_Cluster'] = y_kmeans_test
display(predicted_clusters_df.head())
```

```python
predicted_clusters_df.to_csv('titanic_test_predicted_clusters.csv', index=False)
print('Predicted clusters for test data saved to titanic_test_predicted_clusters.csv')
```
